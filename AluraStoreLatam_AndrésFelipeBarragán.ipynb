## Se analiza los datos de la primera tienda, se evaluo la siguiente información:

### Los ingresos generados por la tienda, las categorías más vendidas en esta tienda, las reseñas de los clientes que compraron en esta tienda, los productos más vendidos en esta tienda y envío promedio realizado en esta tienda

```python
# Información General
import pandas as pd
# Cargar la base de datos desde la URL
url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
df = pd.read_csv(url)
df['Ingresos'] = df['Precio']
df['Reseñas'] = df['Calificación']
# Ingresos totales
ingresos_totales = df['Ingresos'].sum()
print(f"Ingresos totales: ${ingresos_totales:,.2f}")
# Categorias más vendidas
categorias_mas_vendidas = df['Categoría del Producto'].value_counts()
print("\nCategorías más vendidas:")
print(categorias_mas_vendidas)
# Promedio de reseñas por categoria
reseñas_promedio = df.groupby("Categoría del Producto")['Reseñas'].mean()
print("\nReseñas promedio por categoría:")
print(reseñas_promedio)
# Productos más vendidos
top_productos_por_categoria = (
    df.groupby("Categoría del Producto")["Producto"]
      .value_counts()
      .groupby(level=0)
      .head(2)
)
print("\nTop 2 productos más vendidos por categoría:")
print(top_productos_por_categoria)
# Envío promedio
envio_promedio = df['Costo de envío'].mean()
print(f"Envío promedio: ${envio_promedio:.2f}")
```

## Creación de gráficos para visualizar y analizar el comportamiento de las tres tiendas con respecto a las tres métricas: como es el ingreso, reseñas y rendimiento de ventas

### Grafíca de Dispersión para las cuatro tiendas evaluando las métricas como son el ingreso, reseñas y rendimiento de ventas  

```python
import pandas as pd
import matplotlib.pyplot as plt

# URLs raw de los archivos CSV en GitHub
urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_4.csv"
]

def procesar_y_graficar(url, nombre):
    # Leer CSV desde la URL
    df = pd.read_csv(url)

    # Crear columnas
    df["Ingresos"] = df["Precio"] * df["Cantidad de cuotas"]
    df["Reseñas"] = df["Calificación"]
    df["Rendimiento de ventas"] = df["Ingresos"] / df["Cantidad de cuotas"]

    # Scatter plot: Ingresos por categoría
    plt.figure(figsize=(10, 6))
    plt.scatter(df["Categoría del Producto"], df["Ingresos"], alpha=0.7)
    plt.title(f"Ingresos por categoría — {nombre}")
    plt.xlabel("Categoría")
    plt.ylabel("Ingresos")
    plt.xticks(rotation=45, ha="right")
    plt.grid(True)
    plt.show()

    # Scatter plot: Reseñas por categoría
    plt.figure(figsize=(10, 6))
    plt.scatter(df["Categoría del Producto"], df["Reseñas"], alpha=0.7, color="orange")
    plt.title(f"Reseñas por categoría — {nombre}")
    plt.xlabel("Categoría")
    plt.ylabel("Reseñas (calificación)")
    plt.xticks(rotation=45, ha="right")
    plt.grid(True)
    plt.show()

    # Scatter plot: Rendimiento por categoría
    plt.figure(figsize=(10, 6))
    plt.scatter(df["Categoría del Producto"], df["Rendimiento de ventas"], alpha=0.7, color="green")
    plt.title(f"Rendimiento de ventas por categoría — {nombre}")
    plt.xlabel("Categoría")
    plt.ylabel("Rendimiento de ventas")
    plt.xticks(rotation=45, ha="right")
    plt.grid(True)
    plt.show()

# Llamar la función sobre cada URL
for i, url in enumerate(urls, start=1):
    procesar_y_graficar(url, f"Tienda {i}")
```

### Grafíca de Barra para las cuatro tiendas evaluando las métricas como son el ingreso, reseñas y rendimiento de ventas

```python
import pandas as pd
import matplotlib.pyplot as plt

# URLs raw de los archivos CSV en GitHub
urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/main/base-de-datos-challenge1-latam/tienda_4.csv"
]

def procesar_y_graficar(url, nombre):
    # Leer CSV desde la URL
    df = pd.read_csv(url)

    # Crear columnas
    df["Ingresos"] = df["Precio"] * df["Cantidad de cuotas"]
    df["Reseñas"] = df["Calificación"]
    df["Rendimiento de ventas"] = df["Ingresos"] / df["Cantidad de cuotas"]

    # Agrupaciones
    ingresos = df.groupby("Categoría del Producto")["Ingresos"].sum()
    reseñas = df.groupby("Categoría del Producto")["Reseñas"].mean()
    rendimiento = df.groupby("Categoría del Producto")["Rendimiento de ventas"].mean()

    # --- Gráfico 1: Ingresos ---
    plt.figure(figsize=(10, 6))
    ingresos.plot(kind="bar", color="skyblue")
    plt.title(f"Ingresos por Categoría — {nombre}")
    plt.xlabel("Categoría del Producto")
    plt.ylabel("Ingresos totales")
    plt.xticks(rotation=45, ha="right")
    plt.grid(axis='y')
    plt.show()

    # --- Gráfico 2: Reseñas ---
    plt.figure(figsize=(10, 6))
    reseñas.plot(kind="bar", color="orange")
    plt.title(f"Reseñas promedio por Categoría — {nombre}")
    plt.xlabel("Categoría del Producto")
    plt.ylabel("Reseña promedio")
    plt.xticks(rotation=45, ha="right")
    plt.grid(axis='y')
    plt.show()

    # --- Gráfico 3: Rendimiento de ventas ---
    plt.figure(figsize=(10, 6))
    rendimiento.plot(kind="bar", color="green")
    plt.title(f"Rendimiento de Ventas por Categoría — {nombre}")
    plt.xlabel("Categoría del Producto")
    plt.ylabel("Rendimiento promedio")
    plt.xticks(rotation=45, ha="right")
    plt.grid(axis='y')
    plt.show()

# Llamar la función sobre cada URL
for i, url in enumerate(urls, start=1):
    procesar_y_graficar(url, f"Tienda {i}")
```

### Grafíca de Torta para las cuatro tiendas evaluando las métricas como son el ingreso, reseñas y rendimiento de ventas

```python
import matplotlib.pyplot as plt
import pandas as pd

nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
tiendas = [tienda, tienda2, tienda3, tienda4]

# Gráfico de barras agrupadas
# Ventas por categoría en cada tienda

for nombre, df in zip(nombres, tiendas):
    categorias = df["Categoría del Producto"].value_counts()

    plt.figure(figsize=(8,5))
    plt.bar(categorias.index, categorias.values)
    plt.title(f"Ventas por Categoría - {nombre}")
    plt.xlabel("Categoría")
    plt.ylabel("Cantidad de Ventas")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Gáfico de Torta por tienda
# Distribución Porcentual de ventas por categoría

for nombre, df in zip(nombres, tiendas):
    plt.figure(figsize=(7,7))
    conteo = df['Categoría del Producto'].value_counts()
    plt.pie(conteo, labels=conteo.index, autopct="%1.1f%%")
    plt.title(f"Distribución de Ventas por Categoría – {nombre}")
    plt.show()

# Grafico de Línea - Facturación total por tienda
# Precio total por categoría comparada entre tiendas

# Obtener todas las categorías presentes en todas las tiendas
categorias = sorted(set().union(*[df['Categoría del Producto'].unique() for df in tiendas]))

# Crear una gráfica por tienda
for nombre, df in zip(nombres, tiendas):

    # Calcular el precio total por categoría en esta tienda
    precios_cat = []
    for cat in categorias:
        precio = df[df['Categoría del Producto'] == cat]["Precio"].sum()
        precios_cat.append(precio)

    # ---- Gráfica ----
    plt.figure(figsize=(8,5))
    plt.plot(categorias, precios_cat, marker="o")
    plt.title(f"Precio Total por Categoría - {nombre}")
    plt.xlabel("Categoría")
    plt.ylabel("Valor Total ($)")
    plt.grid(True)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

```

```python
import matplotlib.pyplot as plt

tiendas = [tienda, tienda2, tienda3, tienda4]
nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

for datos, nombre in zip(tiendas, nombres):
    plt.figure(figsize=(8,6))

    plt.scatter(
        datos["lon"], datos["lat"],
        s=20, alpha=0.5
    )

    plt.title(f"Distribución geográfica de ventas - {nombre}")
    plt.xlabel("Longitud")
    plt.ylabel("Latitud")
    plt.grid(True)
    plt.show()
```

# Análisis a nivel gráfico

#### El análisis muestra que las Tiendas 1 a 4 presentan una distribución geográfica de ventas prácticamente idéntica, con la mayoría de los puntos concentrados en un clúster principal (Longitud: −76.8 a −72.5, Latitud: 3 a 12), donde se identifican subgrupos de alta, media y baja latitud que reflejan zonas de concentración diferenciadas dentro del mismo mercado. Además, existen dos puntos de venta aislados: uno al este (Longitud ≈ −68, Latitud ≈ 3.8) y otro al sur (Longitud ≈ −70, Latitud ≈ −4), que representan expansiones marginales o posibles operaciones transnacionales. Esta uniformidad indica que las cuatro tiendas operan sobre el mismo conjunto de datos geográficos o un mercado centralizado, lo que limita la diferenciación geográfica en su análisis de ventas.

```python
for datos, nombre in zip(tiendas, nombres):
    plt.figure(figsize=(8,6))

    plt.hexbin(
        datos["lon"], datos["lat"],
        gridsize=30,
        cmap="inferno"
    )

    plt.colorbar(label="Cantidad de ventas")
    plt.title(f"Heatmap de densidad de ventas - {nombre}")
    plt.xlabel("Longitud")
    plt.ylabel("Latitud")
    plt.show()
```

# Análisis a nivel gráfico

#### El análisis del Heatmap de densidad de ventas de las tiendas 1 a 4 evidencia que la actividad comercial está altamente concentrada en tres Hotspots principales: el Punto Caliente 1 (≈-75.8, 6.5) concentra la mayor densidad con 650-800+ ventas, el Punto Caliente 2 (≈-74.2, 4.5) registra entre 400-600 ventas, y el Punto Caliente 3 (≈-76.2, 3.5) aporta un volumen medio de 200-400 ventas, mientras que el resto del territorio, representado mayormente en negro, muestra ventas prácticamente nulas. Esta distribución agrupada indica que las cuatro tiendas operan sobre un mismo mercado o que los datos están agregados, con una fuerte polarización entre áreas de alta y baja actividad. Estratégicamente, esto implica que los esfuerzos de marketing, promociones, asignación de inventario y logística deben priorizar los Hotspots principales para maximizar eficiencia y rentabilidad, mientras que las zonas circundantes con baja densidad presentan oportunidades de expansión limitada, aunque el riesgo de depender de tan pocos focos de venta es elevado ante cambios de mercado o competencia.


```python
import folium
from folium.plugins import HeatMap

tiendas = [tienda, tienda2, tienda3, tienda4]
nombres = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]

for datos, nombre in zip(tiendas, nombres):

    # centro del mapa
    lat_centro = datos["lat"].mean()
    lon_centro = datos["lon"].mean()

    mapa = folium.Map(location=[lat_centro, lon_centro], zoom_start=12)

    HeatMap(datos[["lat", "lon"]].values).add_to(mapa)

    mapa.save(f"heatmap_{nombre}.html")
    print(f"Mapa generado: heatmap_{nombre}.html")
```

### Subtask:
Unir todos los DataFrames de las tiendas (tienda, tienda2, tienda3, tienda4) en un único DataFrame para facilitar el análisis.


**Reasoning**:
To combine the four DataFrames into a single one, I will use the `pd.concat` function, which stacks them row-wise.



```python
df_combined = pd.concat([tienda, tienda2, tienda3, tienda4], ignore_index=True)
df_combined.head()
```

**Reasoning**:
Now that the DataFrames are combined, I need to extract all unique product categories from the `df_combined` DataFrame to prepare for creating category-specific plots, as outlined in the task instructions.



```python
unique_categories = df_combined['Categoría del Producto'].unique()
print("Unique Product Categories:")
for category in unique_categories:
    print(category)
```

**Reasoning**:
Now that the unique categories have been extracted, the next step is to generate a plot for each of these categories. This will involve filtering the `df_combined` DataFrame by each unique category, selecting only the numerical columns from the filtered data, and then plotting these numerical columns.



```python
import matplotlib.pyplot as plt

for category in unique_categories:
    category_df = df_combined[df_combined['Categoría del Producto'] == category]
    # Select only numerical columns for plotting, excluding lat/lon as they are coordinates
    numerical_cols = category_df.select_dtypes(include=['number']).drop(columns=['lat', 'lon'], errors='ignore')

    if not numerical_cols.empty:
        plt.figure(figsize=(10, 6))
        numerical_cols.plot(title=f'Numerical Data for {category} Category')
        plt.tight_layout()
        plt.show()
    else:
        print(f"No numerical data to plot for {category} category (excluding lat/lon).")
```
