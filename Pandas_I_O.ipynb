### **Actualizando un banco de datos**

```python
query2 = 'SELECT * FROM clientes'
```

```python
pd.read_sql(sql=text(query2),con = engine.connect())
```

```python
# Borrando registros
from sqlalchemy.exc import SQLAlchemyError
query3 = "DELETE FROM clientes WHERE ID_Cliente = 5008804"
try:
  with engine.connect() as conn:
    r_set = conn.execute(text(query3))
    conn.commit()
except SQLAlchemyError as e:
  print(e)
else:
  print("#Registros borrados: ", r_set.rowcount)
```

```python
pd.read_sql_table('clientes',con = engine.connect())
```

```python
# Actualizado registros
query4 = "UPDATE clientes SET Grado_estudio='Nivel superior' WHERE ID_Cliente = 5008808"
try:
  with engine.connect() as conn:
    r_set = conn.execute(text(query4))
    conn.commit()
except SQLAlchemyError as e:
  print(e)
else:
  print("#Registros actualizados: ", r_set.rowcount)
```

```python
pd.read_sql_table('clientes',con = engine.connect())
```

```python
inspector2 = inspect(engine)
print(inspector2.get_table_names())
```

#**Desafío: manipulando una base de datos**

### **Impotamos la base de datos**

### **Crear la base de datos local con la biblioteca SQLAlchemy**

```python
import sqlalchemy
from sqlalchemy import create_engine, MetaData, Table, inspect, text
engine2 = create_engine('sqlite:///:memory:')
```

### **Escribir los datos del archivo CSV en esta base de datos local**

```python
import pandas as pd
archivo = 'clientes_banco.csv'
datos_banco = pd.read_csv(archivo)
datos_banco.to_sql('clientes',engine2,index=False)
```

```python
pd.read_sql_table('clientes',con = engine2.connect())
```

### **Realizar tres actualizaciones en la base de datos:**

#### **Actualizar el registro del cliente ID 6840104 cuyo rendimiento anual cambió a 300000.**

```python
from sqlalchemy.exc import SQLAlchemyError
query5 = 'UPDATE clientes SET Rendimiento_anual = 300000.0 WHERE ID_Cliente = 6840104'
try:
  r_set = engine2.connect().execute(text(query5))
except SQLAlchemyError as e:
  print(e)
else:
  print("#Registros actualizados: ", r_set.rowcount)
```

```python
pd.read_sql_table('clientes',con = engine2.connect())
```

#### **Eliminar el registro de cliente ID 5008809, ya que esta persona ya no tiene cuenta en la institución financiera.**

```python
from sqlalchemy.exc import SQLAlchemyError
query6 = 'DELETE FROM clientes WHERE ID_Cliente = 5008809'
try:
  r_set = engine2.connect().execute(text(query6))
except SQLAlchemyError as e:
  print(e)
else:
  print("#Registros borrados: ", r_set.rowcount)
```

```python
pd.read_sql_table('clientes',con = engine2.connect())
```

### **Crea un nuevo registro de cliente siguiendo las especificaciones a continuación:**

#### **ID_cliente: 6850985**
#### **Edad: 33**
#### **Grado_estudio: Doctorado**
#### **Estado_civil: Soltero**
#### **Tamaño_familia: 1**
#### **Categoria_de_renta: Empleado**
#### **Ocupacion: TI**
#### **Años_empleado: 2**
#### **Rendimiento_anual: 290000**
#### **Tiene_carro: 0**
#### **Vivienda: Casa/Departamento propio**

```python
from sqlalchemy.exc import SQLAlchemyError
query7 = 'INSERT INTO clientes (ID_Cliente, Edad, Grado_estudio, Estado_civil, ' \
          'Tamaño_familia, Categoria_de_renta, Ocupacion, Años_empleado, ' \
          'Rendimiento_anual, Tiene_carro, Vivienda) ' \
          'VALUES (6850985, 33, "Doctorado", "Soltero", 1, "Empleado", "TI", 2, 290000, 0, "Casa/Departamento propio")'
try:
  r_set = engine2.connect().execute(text(query7))
except SQLAlchemyError as e:
  print(e)
else:
  print("#Registros insertados: ", r_set.rowcount)
```

```python
pd.read_sql_table('clientes',con = engine2.connect())
```
